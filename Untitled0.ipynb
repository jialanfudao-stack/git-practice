# Colabで文章Qwenと画像Qwenを同時起動

指定されたHuihui量子化済みGGUFをGoogle ColabのA100上でllama.cppに読み込みます。serverはColab内の127.0.0.1だけに起動し、通常はipywidgetsから呼び出します。VS Codeから使う場合だけ、後半の任意ボタンで認証付き非stream proxyを起動し、Cloudflare Quick Tunnelへ接続します。

モデル: [huihui-ai/Huihui-Qwen3.8-27B-abliterated-GGUF](https://huggingface.co/huihui-ai/Huihui-Qwen3.8-27B-abliterated-GGUF)
- 使用ファイルは Huihui-Qwen3.8-27B-abliterated-Q6_K_L.gguf（約24.9GB）です。Q6_K_Lは通常のllama.cppで扱う混合量子化GGUFで、Ternary/Prism専用形式ではありません。
- ベースはQwen/Qwen3.8-27B、アーキテクチャはqwen35です。MTPと視覚部分は変更されていませんが、このノートブックではHuihuiを文章用serverとして、Qwen-Image-2.1を画像生成用ComfyUIとして起動します。MTPは初期無効です。

参照: [Huihuiモデルカード](https://huggingface.co/huihui-ai/Huihui-Qwen3.8-27B-abliterated-GGUF) / [llama.cpp CUDAビルド](https://github.com/ggml-org/llama.cpp/blob/master/docs/build.md) / [server API](https://github.com/ggml-org/llama.cpp/blob/master/tools/server/README.md) / [固定commit](https://github.com/ggml-org/llama.cpp/commit/030ebb558a5820b444a8f836ed5cdd46c9b4bd7a)

## 実行順（A100 40GB）

1. ColabランタイムをA100にします。
2. このノートブックを上から実行します。最初にHuihuiを起動し、続いて画像モデルを準備し、cell21でComfyUIを自動起動します。画像の重みを追加で取得します。
3. 「チャットUI」「画像モードUI」の両方が表示され、二つのlocalhost APIが稼働したら準備完了です。ノートブックUIからのチャットと画像生成はGPU共有ロックで順番に実行します。
4. VS Code接続を使う場合は、最後の「VS Code接続を開始」を押し、URLとモデルだけを確認して設定JSONをPCへダウンロードします。Continueでは文章モデル `huihui-qwen38-abliterated-q6kl-gguf` を選択します。画像生成用ComfyUIのモデルは別APIなのでチャットには選びません。応答は非stream、最大128 tokenを初期値にします。
5. 終了時は「両バックエンドを停止」を押してからColabランタイムを切断します。

両server processを維持しますが、Q6/Q8の全重みをA100 40GBへ同時に載せる構成ではありません。Huihuiは一部レイヤーをGPUへ置き、残りをRAMへ置くため、起動前に利用可能なCPU RAM 32GiB以上を確認します。ComfyUIは低VRAMモードを使います。ノートブックUIと認証付きVS Code bridgeはGPU共有ロックを使います。その他の直接API呼び出しは共有ロックを使いません。

同じColab VMでセルを再実行するとき、検証stampと実ファイルが一致するモデル・build・依存は再利用します。`/content` はColab VM内の一時領域です。Runtime の Restart session はVM上のファイルを残してカーネルを再起動しますが、Disconnect and delete runtime はVMとファイルを削除します。Driveへの自動保存はしません。

この更新を既存のColabランタイムに反映するときは、ランタイムを削除せず、更新版Notebookを上から実行してください。検証stampと実ファイルが有効なモデル・build・依存は再利用されます。

```python
import hashlib, json
from huggingface_hub import HfApi, hf_hub_download
REPO_ID = 'huihui-ai/Huihui-Qwen3.8-27B-abliterated-GGUF'
MODEL_FILENAME = 'Huihui-Qwen3.8-27B-abliterated-Q6_K_L.gguf'
MODEL_REVISION = '3f101cd22b7999228bbd5d79a33975414eb9758b'
EXPECTED_SHA256 = '20b1214a23f0f5f401b6117cbd63336cfff6c07be4df93bd6e374d8fd8d12938'
MODEL_DIR = Path('/content/qwen38_gguf'); MODEL_DIR.mkdir(parents=True, exist_ok=True)
model_path = MODEL_DIR / MODEL_FILENAME
model_stamp = MODEL_DIR / (MODEL_FILENAME + '.verified.json')

def _sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for block in iter(lambda: handle.read(16 * 1024 * 1024), b''): digest.update(block)
    return digest.hexdigest().lower()

def _model_stamp_valid():
    try:
        stat = model_path.stat()
        stamp = json.loads(model_stamp.read_text(encoding='utf-8'))
        return (stat.st_size > 0 and stamp == {
            'revision': MODEL_REVISION, 'sha256': EXPECTED_SHA256,
            'size': stat.st_size, 'mtime_ns': stat.st_mtime_ns,
        })
    except (OSError, ValueError, TypeError):
        return False

model_reused = _model_stamp_valid()
if not model_reused:
    if not re.fullmatch(r'[0-9a-f]{40}', MODEL_REVISION): raise RuntimeError(f'revision pinが不正です: {MODEL_REVISION!r}')
    if not re.fullmatch(r'[0-9a-f]{64}', EXPECTED_SHA256): raise RuntimeError('期待SHA256の形式が不正です')
    try:
        info = HfApi().model_info(REPO_ID, revision=MODEL_REVISION)
    except Exception as exc:
        raise RuntimeError('Hugging Faceから固定revisionを確認できませんでした。') from exc
    if info.sha != MODEL_REVISION: raise RuntimeError(f'revision不一致: expected={MODEL_REVISION}, actual={info.sha}')
    if model_path.is_file() and model_path.stat().st_size > 0 and _sha256_file(model_path) == EXPECTED_SHA256:
        print('既存HuihuiをSHA256検証しました。ダウンロードを省略します:', model_path)
    else:
        free_gib = shutil.disk_usage('/content').free / (1024**3)
        if free_gib < 50:
            raise RuntimeError(f'モデル再取得に必要な空き容量がありません: {free_gib:.1f}GiB < 50GiB')
        model_path = Path(hf_hub_download(repo_id=REPO_ID, filename=MODEL_FILENAME, revision=MODEL_REVISION, local_dir=MODEL_DIR, force_download=True))
        if model_path.name != MODEL_FILENAME:
            raise RuntimeError(f'取得したモデル名が一致しません: {model_path.name}')
        if not model_path.is_file() or model_path.stat().st_size <= 0:
            raise RuntimeError(f'モデルファイルが空か存在しません: {model_path}')
    actual_sha256 = _sha256_file(model_path)
    if actual_sha256 != EXPECTED_SHA256: raise RuntimeError(f'SHA256不一致: expected={EXPECTED_SHA256}, actual={actual_sha256}')
    stat = model_path.stat()
    model_stamp.write_text(json.dumps({'revision': MODEL_REVISION, 'sha256': EXPECTED_SHA256, 'size': stat.st_size, 'mtime_ns': stat.st_mtime_ns}), encoding='utf-8')
else:
    print('既存Huihuiはrevision/SHA256検証stampとファイル属性が一致。再検証・取得を省略します:', model_path)
print(f'Huihui準備・SHA256検証OK: {model_path} ({model_path.stat().st_size/(1024**3):.2f}GiB), revision={MODEL_REVISION}')
```

## localhostサーバー

Huihuiは llama.cpp、画像生成はComfyUIのlocalhost APIを使います。A100 40GB内に両モデルの全重みは収まらないため、HuihuiのGPUレイヤー数とコンテキスト長を制限し、ComfyUIは低VRAM/CPUオフロードで起動します。画像生成中も二つのserver processとhealth endpointを維持します。

```python
import time, requests, socket, subprocess, threading, re
from pathlib import Path

# パスを自動解決（前のセルが未実行でもファイルを直接参照）
SERVER_BIN = globals().get('SERVER_BIN') or Path('/content/qwen38_work/llama.cpp/build/bin/llama-server')
model_path = globals().get('model_path') or Path('/content/qwen38_gguf/Huihui-Qwen3.8-27B-abliterated-Q6_K_L.gguf')

# ファイルの存在確認
if not SERVER_BIN.exists():
    raise RuntimeError(f"llama-serverがありません。上のビルドセル（cmakeコマンド等があるセル）を実行してください。")
if not model_path.exists():
    raise RuntimeError(f"モデルファイルがありません。上のダウンロードセルを実行してください。")

LLAMA_GPU_LAYERS = 16
LLAMA_CTX_SIZE = 4096
MIN_LLM_HOST_RAM_GIB = 32
SERVER_HOST = '127.0.0.1'
MODEL_ALIAS = 'huihui-qwen38-abliterated-q6kl-gguf'
ENABLE_MTP = False

server_process = globals().get('server_process')
server_log_handle = globals().get('server_log_handle')

def _matches_tracked_huihui(proc):
    if proc is None or proc.poll() is not None:
        return False
    try:
        import psutil
        argv = psutil.Process(proc.pid).cmdline()
        return (
            Path(argv[0]).resolve() == Path(str(SERVER_BIN)).resolve()
            and Path(argv[argv.index('-m') + 1]).resolve() == Path(str(model_path)).resolve()
            and argv[argv.index('--alias') + 1] == MODEL_ALIAS
        )
    except Exception:
        return False

def stop_server():
    global server_process, server_log_handle
    if server_process is not None and server_process.poll() is None:
        if not _matches_tracked_huihui(server_process):
            raise RuntimeError('追跡中のserver processがHuihuiと一致しないため停止しません。')
        server_process.terminate()
        try: server_process.wait(timeout=20)
        except subprocess.TimeoutExpired:
            server_process.kill(); server_process.wait(timeout=10)
    server_process = None
    if server_log_handle is not None: server_log_handle.close(); server_log_handle = None

# Keep an already-running, matching server. Never terminate an unknown process.
if server_process is not None and server_process.poll() is None:
    import psutil
    running_argv = psutil.Process(server_process.pid).cmdline()
    if not _matches_tracked_huihui(server_process):
        raise RuntimeError('追跡中のserver processがHuihuiと一致しないため、既存プロセスを停止しません。')
    current_base = str(globals().get('BASE_URL', '')).rstrip('/')
    if current_base and requests.get(f'{current_base}/health', timeout=5).status_code == 200:
        BASE_URL = current_base
        print('既存のHuihui serverを再利用:', BASE_URL)
        REUSE_HUIHUI = True
    else:
        raise RuntimeError('既存のHuihui processがまだ準備できていません。')
else:
    REUSE_HUIHUI = False
    import psutil
    host_ram_gib = psutil.virtual_memory().available / (1024**3)
    if host_ram_gib < MIN_LLM_HOST_RAM_GIB:
        raise RuntimeError(f'CPU RAM不足: 空き{host_ram_gib:.1f}GiB。32GiB以上を確保してください。RAM不足の場合はColabの高RAMランタイムを選んでください。')
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as port_probe:
        port_probe.bind((SERVER_HOST, 0))
        SERVER_PORT = port_probe.getsockname()[1]

    BASE_URL = f'http://{SERVER_HOST}:{SERVER_PORT}'
    server_cmd = [str(SERVER_BIN),'-m',str(model_path),'--alias',MODEL_ALIAS,'--host',SERVER_HOST,'--port',str(SERVER_PORT),'--no-webui','--jinja','--n-gpu-layers',str(LLAMA_GPU_LAYERS),'--ctx-size',str(LLAMA_CTX_SIZE),'--parallel','1','--chat-template-kwargs','{"enable_thinking":false}',]
    if ENABLE_MTP: server_cmd += ['--spec-type','draft-mtp','--spec-draft-n-max','3']

    server_log_handle = open('/content/llama_server.log', 'a', encoding='utf-8')
    server_process = subprocess.Popen(server_cmd, stdout=server_log_handle, stderr=subprocess.STDOUT, text=True)

print("llama-server 起動待機中...")
deadline = time.monotonic()+900
while time.monotonic()<deadline:
    if server_process.poll() is not None:
        try:
            tail = Path('/content/llama_server.log').read_text(encoding='utf-8', errors='replace').splitlines()[-40:]
            safe_tail = [line for line in tail if not re.search(r'(?i)(token|secret|password|authorization)', line)]
            print('llama-server起動ログ（秘密情報らしい行は除外）:\n' + '\n'.join(safe_tail)[-4000:])
        except Exception:
            pass
        stop_server()
        raise RuntimeError('llama-serverが終了しました。')
    try:
        if requests.get(f'{BASE_URL}/health', timeout=3).status_code == 200: break
    except requests.RequestException: pass
    time.sleep(3)
else:
    stop_server()
    raise TimeoutError('サーバー起動が制限時間内に完了しませんでした。ログ: /content/llama_server.log')

model_ids = [str(x.get('id')) for x in requests.get(f'{BASE_URL}/v1/models', timeout=10).json().get('data', [])]
if MODEL_ALIAS not in model_ids:
    raise RuntimeError(f'Huihui aliasが見つかりません: {model_ids}')
print('localhostサーバー起動OK:', BASE_URL)
print(f'配置設定: GPU layers={LLAMA_GPU_LAYERS}, context={LLAMA_CTX_SIZE}, CPU RAM空き={psutil.virtual_memory().available/(1024**3):.1f}GiB')

print('Huihuiはlocalhost限定です。公開トンネルは起動しません。')
```

```python
import threading
GPU_INFERENCE_LOCK = globals().get('GPU_INFERENCE_LOCK') or threading.Lock()
GPU_INFERENCE_BLOCKED = globals().get('GPU_INFERENCE_BLOCKED', False)
def run_gpu_serialized(fn, *args, **kwargs):
    if not GPU_INFERENCE_LOCK.acquire(blocking=False):
        raise RuntimeError('GPUは別の推論で使用中です。少し待ってからもう一度お試しください。')
    try:
        if GPU_INFERENCE_BLOCKED:
            raise RuntimeError('推論状態を確認できません。両backendを停止してから再起動してください。')
        return fn(*args, **kwargs)
    finally:
        GPU_INFERENCE_LOCK.release()

SYSTEM_PROMPT = 'あなたは親切な日本語アシスタントです。簡潔で正確に答えてください。'
MAX_HISTORY_TURNS, MAX_HISTORY_CHARS, MAX_NEW_TOKENS = 8, 24000, 512
PROMPT_SAFETY_TOKENS = 64
def prompt_token_budget(context_size, max_new_tokens, safety_tokens=PROMPT_SAFETY_TOKENS):
    if context_size < 1 or max_new_tokens < 1 or max_new_tokens + safety_tokens >= context_size:
        raise ValueError('context size must leave room for prompt, output, and safety tokens')
    return context_size - max_new_tokens - safety_tokens
MAX_PROMPT_TOKENS = prompt_token_budget(LLAMA_CTX_SIZE, MAX_NEW_TOKENS)
chat_history = []
def _trim_history(history):
    pairs=[]; pending=None
    for item in history:
        if item.get('role')=='user': pending=item
        elif item.get('role')=='assistant' and pending is not None: pairs.append((pending,item)); pending=None
    kept=[]; total=0
    for pair in reversed(pairs[-MAX_HISTORY_TURNS:]):
        chars=sum(len(str(x.get('content',''))) for x in pair)
        if total+chars>MAX_HISTORY_CHARS: break
        kept.append(pair); total+=chars
    return [item for pair in reversed(kept) for item in pair]
def _messages(user_text, history): return [{'role':'system','content':SYSTEM_PROMPT}, *history, {'role':'user','content':user_text}]
def _template_token_count(messages):
    r=requests.post(f'{BASE_URL}/apply-template',json={'messages':messages},timeout=30); r.raise_for_status(); prompt=r.json().get('prompt')
    if not isinstance(prompt,str): raise RuntimeError('/apply-templateのpromptがありません。')
    r=requests.post(f'{BASE_URL}/tokenize',json={'content':prompt,'add_special':False},timeout=30); r.raise_for_status(); tokens=r.json().get('tokens')
    if not isinstance(tokens,list): raise RuntimeError('/tokenizeのtokensがありません。')
    return len(tokens)
def generate_reply(user_text, history=None, max_tokens=MAX_NEW_TOKENS):
    user_text=user_text.strip()
    if not user_text: raise ValueError('質問を入力してください。')
    prompt_budget = prompt_token_budget(LLAMA_CTX_SIZE, max_tokens)
    history=_trim_history(history or [])
    while True:
        messages=_messages(user_text,history); prompt_tokens=_template_token_count(messages)
        if prompt_tokens<=prompt_budget: break
        if history: history=history[2:]; continue
        raise ValueError(f'入力が長すぎます（テンプレート後{prompt_tokens}トークン）。質問を短くして、{MAX_PROMPT_TOKENS}トークン以内にしてください。')
    payload={'model':MODEL_ALIAS,'messages':messages,'max_tokens':max_tokens,'temperature':0.7,'top_p':0.8,'reasoning_effort':'none','chat_template_kwargs':{'enable_thinking':False}}
    try:
        r=requests.post(f'{BASE_URL}/v1/chat/completions',json=payload,timeout=900)
        r.raise_for_status()
    except requests.RequestException as exc:
        global GPU_INFERENCE_BLOCKED
        GPU_INFERENCE_BLOCKED = True
        raise RuntimeError('文章serverの応答状態を確認できません。両backendを停止してから再起動してください。') from exc
    data=r.json()
    try: answer=data['choices'][0]['message']['content']
    except (KeyError,IndexError,TypeError) as exc: raise RuntimeError('chat completions応答形式が不正です。') from exc
    return str(answer).strip()
r=requests.get(f'{BASE_URL}/v1/models',timeout=10); r.raise_for_status(); print('API接続OK:', r.json().get('data',[{}])[0].get('id',MODEL_ALIAS))
```

```python
try:
    from google.colab import output as colab_output
    colab_output.enable_custom_widget_manager()
except ImportError:
    pass

import ipywidgets as widgets
from IPython.display import clear_output, display

def create_chat_ui(base_url, model_alias, initial_history=None):
    """Create a chat UI whose callbacks keep one verified server endpoint."""
    base_url = str(base_url).rstrip('/')
    model_alias = str(model_alias)
    history = list(initial_history or [])

    def check_server():
        health = requests.get(f'{base_url}/health', timeout=10)
        health.raise_for_status()
        models_response = requests.get(f'{base_url}/v1/models', timeout=10)
        models_response.raise_for_status()
        model_ids = [str(item.get('id')) for item in models_response.json().get('data', [])]
        if model_alias not in model_ids:
            raise RuntimeError(f'接続先のモデルaliasが一致しません: expected={model_alias}, actual={model_ids}')

    def trim_history(items):
        pairs = []
        pending = None
        for item in items:
            if item.get('role') == 'user':
                pending = item
            elif item.get('role') == 'assistant' and pending is not None:
                pairs.append((pending, item))
                pending = None
        kept = []
        total = 0
        for pair in reversed(pairs[-MAX_HISTORY_TURNS:]):
            chars = sum(len(str(item.get('content', ''))) for item in pair)
            if total + chars > MAX_HISTORY_CHARS:
                break
            kept.append(pair)
            total += chars
        return [item for pair in reversed(kept) for item in pair]

    def token_count(messages):
        response = requests.post(
            f'{base_url}/apply-template',
            json={'messages': messages},
            timeout=30,
        )
        response.raise_for_status()
        prompt = response.json().get('prompt')
        if not isinstance(prompt, str):
            raise RuntimeError('/apply-templateのpromptがありません。')
        response = requests.post(
            f'{base_url}/tokenize',
            json={'content': prompt, 'add_special': False},
            timeout=30,
        )
        response.raise_for_status()
        tokens = response.json().get('tokens')
        if not isinstance(tokens, list):
            raise RuntimeError('/tokenizeのtokensがありません。')
        return len(tokens)

    def _generate_unlocked(user_text, previous=None, max_tokens=MAX_NEW_TOKENS):
        user_text = user_text.strip()
        if not user_text:
            raise ValueError('質問を入力してください。')
        check_server()
        prompt_budget = prompt_token_budget(LLAMA_CTX_SIZE, max_tokens)
        previous = trim_history(previous or [])
        while True:
            messages = [
                {'role': 'system', 'content': SYSTEM_PROMPT},
                *previous,
                {'role': 'user', 'content': user_text},
            ]
            prompt_tokens = token_count(messages)
            if prompt_tokens <= prompt_budget:
                break
            if previous:
                previous = previous[2:]
                continue
            raise ValueError(
                f'入力が長すぎます（テンプレート後{prompt_tokens}トークン）。'
                f'質問を短くして、出力{max_tokens}トークン分を確保した{prompt_budget}トークン以内にしてください。'
            )
        payload = {
            'model': model_alias,
            'messages': messages,
            'max_tokens': max_tokens,
            'temperature': 0.7,
            'top_p': 0.8,
            'reasoning_effort': 'none',
            'chat_template_kwargs': {'enable_thinking': False},
        }
        try:
            response = requests.post(
                f'{base_url}/v1/chat/completions',
                json=payload,
                timeout=900,
            )
            response.raise_for_status()
        except requests.RequestException as exc:
            global GPU_INFERENCE_BLOCKED
            GPU_INFERENCE_BLOCKED = True
            raise RuntimeError('文章serverの応答状態を確認できません。両backendを停止してから再起動してください。') from exc
        data = response.json()
        try:
            answer = data['choices'][0]['message']['content']
        except (KeyError, IndexError, TypeError) as exc:
            raise RuntimeError('chat completions応答形式が不正です。') from exc
        return str(answer).strip()

    def generate(user_text, previous=None, max_tokens=MAX_NEW_TOKENS):
        return run_gpu_serialized(_generate_unlocked, user_text, previous, max_tokens)

    prompt_box = widgets.Textarea(
        placeholder='質問を入力してください',
        layout=widgets.Layout(width='100%', height='90px'),
    )
    send_button = widgets.Button(description='送信', button_style='primary')
    reset_button = widgets.Button(description='会話をリセット')
    chat_output = widgets.Output(layout={'border': '1px solid #ddd'})

    def send(_):
        text = prompt_box.value.strip()
        if not text:
            return
        send_button.disabled = True
        try:
            answer = generate(text, previous=history)
            history[:] = trim_history(
                history + [
                    {'role': 'user', 'content': text},
                    {'role': 'assistant', 'content': answer},
                ]
            )
            chat_history[:] = history
            with chat_output:
                print(f'あなた: {text}')
                print(f'Huihui Qwen3.8 Q6_K_L: {answer}')
                print()
            prompt_box.value = ''
        except Exception as exc:
            with chat_output:
                print(f'エラー: {type(exc).__name__}')
        finally:
            prompt_box.value = ''
            send_button.disabled = False

    def reset(_):
        history.clear()
        chat_history.clear()
        prompt_box.value = ''
        with chat_output:
            clear_output(wait=True)
            print('会話をリセットしました。')

    check_server()
    send_button.on_click(send)
    reset_button.on_click(reset)
    return prompt_box, send_button, reset_button, chat_output

clear_output(wait=True)
prompt_box, send_button, reset_button, chat_output = create_chat_ui(BASE_URL, MODEL_ALIAS)
display(widgets.VBox([prompt_box, widgets.HBox([send_button, reset_button]), chat_output]))
print('HUIHUI_CHAT_UI_READY', BASE_URL, MODEL_ALIAS)
```

```python

# Explicit minimal API graphs. Validate node names and inputs against /object_info after ComfyUI starts.
import io, json, threading, uuid, requests
from IPython.display import display

IMAGE_HOST = '127.0.0.1'
IMAGE_MODEL_ALIAS = 'qwen-image-2.1-uc-q8'
IMAGE_PORT = globals().get('IMAGE_PORT')
IMAGE_BASE_URL = globals().get('IMAGE_BASE_URL')
IMAGE_SERVER_PROCESS = globals().get('IMAGE_SERVER_PROCESS')
IMAGE_SERVER_LOG_HANDLE = globals().get('IMAGE_SERVER_LOG_HANDLE')
IMAGE_LAST_OUTPUT = globals().get('IMAGE_LAST_OUTPUT')

def validate_image_object_info(base_url):
    info = requests.get(f'{base_url}/object_info', timeout=30).json()
    required = {'UnetLoaderGGUF', 'CLIPLoader', 'VAELoader', 'TextEncodeQwenImage21', 'QwenImage21Cache', 'EmptyLatentImage', 'KSampler', 'VAEDecode', 'QwenEncryptedSaveImage', 'QwenMemoryReference'}
    missing = sorted(required - set(info))
    if missing:
        raise RuntimeError(f'ComfyUI object_infoに必要ノードがありません: {missing}')
    for cls, names in {
        'TextEncodeQwenImage21': {'clip', 'prompt', 'negative_prompt', 'resolution'},
        'QwenImage21Cache': {'model', 'device', 'dtype'},
    }.items():
        inputs = info[cls].get('input', {})
        names_present = set(inputs.get('required', {})) | set(inputs.get('optional', {}))
        if not names <= names_present:
            raise RuntimeError(f'{cls}の入力schemaが想定外です: {sorted(names_present)}')
    print('object_info schema OK:', ', '.join(sorted(required)))

def build_t2i_graph(prompt, width=1024, height=1024, seed=20261001, steps=25):
    return {
        '1': {'class_type': 'UnetLoaderGGUF', 'inputs': {'unet_name': IMAGE_GGUF}},
        '2': {'class_type': 'CLIPLoader', 'inputs': {'clip_name': Path(IMAGE_TE).name, 'type': 'qwen_image'}},
        '3': {'class_type': 'VAELoader', 'inputs': {'vae_name': Path(IMAGE_VAE).name}},
        '4': {'class_type': 'TextEncodeQwenImage21', 'inputs': {'clip': ['2', 0], 'prompt': prompt, 'negative_prompt': '', 'resolution': max(width, height)}},
        '5': {'class_type': 'QwenImage21Cache', 'inputs': {'model': ['1', 0], 'device': 'auto', 'dtype': 'default'}},
        '6': {'class_type': 'EmptyLatentImage', 'inputs': {'width': width, 'height': height, 'batch_size': 1}},
        '7': {'class_type': 'KSampler', 'inputs': {'model': ['5', 0], 'positive': ['4', 0], 'negative': ['4', 1], 'latent_image': ['6', 0], 'seed': seed, 'steps': steps, 'cfg': 1.0, 'sampler_name': 'euler', 'scheduler': 'simple', 'denoise': 1.0}},
        '8': {'class_type': 'VAEDecode', 'inputs': {'samples': ['7', 0], 'vae': ['3', 0]}},
        '9': {'class_type': 'QwenEncryptedSaveImage', 'inputs': {'image': ['8', 0], 'public_key_path': str(PRIVATE_PUBLIC_KEY_PATH), 'result_id': f'qwen-image-{uuid.uuid4().hex}'}},
    }

def build_i2i_graph(ref_id, prompt, width=1024, height=1024, seed=20261001, steps=25):
    graph = build_t2i_graph(prompt, width, height, seed, steps)
    graph['10'] = {'class_type': 'QwenMemoryReference', 'inputs': {'ref_id': ref_id}}
    graph['4']['inputs'].update({'images.image_1': ['10', 0], 'vae': ['3', 0]})
    graph['7']['inputs']['latent_image'] = ['4', 2]
    return graph

def queue_image_graph(graph, base_url=None):
    base_url = base_url or IMAGE_BASE_URL
    if not base_url:
        raise RuntimeError('画像backendが起動していません。')
    payload = {'prompt': graph, 'client_id': str(uuid.uuid4())}
    response = requests.post(f'{base_url}/prompt', json=payload, timeout=30)
    response.raise_for_status()
    return response.json()['prompt_id']

def clear_comfy_history(base_url, prompt_id):
    try:
        response = requests.post(
            f'{base_url}/history',
            json={'delete': [str(prompt_id)]},
            timeout=10,
        )
        response.raise_for_status()
        return True
    except requests.RequestException:
        return False

def wait_image_result(prompt_id, base_url=None, timeout=900):
    base_url = base_url or IMAGE_BASE_URL
    if not base_url:
        raise RuntimeError('画像backendが起動していません。')
    deadline = time.monotonic() + timeout
    result = None
    try:
        while time.monotonic() < deadline:
            response = requests.get(f'{base_url}/history/{prompt_id}', timeout=30)
            response.raise_for_status()
            history = response.json().get(prompt_id)
            if history and history.get('status', {}).get('status_str') == 'error':
                raise RuntimeError('画像生成に失敗しました。詳細な履歴は表示しません。')
            if history and history.get('outputs'):
                for node in history['outputs'].values():
                    for item in node.get('private_result', []):
                        result = item
                        break
                    if result is not None:
                        break
            if result is not None:
                break
            time.sleep(2)
        if result is None:
            raise TimeoutError('画像生成が制限時間内に完了しませんでした。')
        return result, None
    finally:
        if not clear_comfy_history(base_url, prompt_id):
            raise RuntimeError('生成履歴の消去に失敗しました。結果は表示していません。')
```

```python
# Start ComfyUI while retaining the notebook-owned Huihui server on localhost.
import re, socket, subprocess, time
from pathlib import Path
IMAGE_PORT = globals().get('IMAGE_PORT')
IMAGE_BASE_URL = globals().get('IMAGE_BASE_URL')
IMAGE_SERVER_PROCESS = globals().get('IMAGE_SERVER_PROCESS')
IMAGE_SERVER_LOG_HANDLE = globals().get('IMAGE_SERVER_LOG_HANDLE')

IMAGE_MIN_FREE_VRAM_GIB = 20.0

def _gpu_free_gib():
    try:
        fields = subprocess.run(['nvidia-smi', '--query-gpu=memory.total,memory.used', '--format=csv,noheader,nounits'], check=True, capture_output=True, text=True).stdout.strip().splitlines()[0].split(',')
        total_mib, used_mib = (int(float(x.strip())) for x in fields)
        return max(0.0, (total_mib - used_mib) / 1024)
    except Exception as exc:
        raise RuntimeError('GPU空きVRAMを確認できないため、画像モデルを起動しません。') from exc

def _matches_comfy_process(proc):
    try:
        argv = list(proc.args)
        return len(argv) > 1 and Path(str(argv[1])).resolve() == Path(str(IMAGE_COMFY_DIR / 'main.py')).resolve()
    except Exception:
        return False

def _clear_image_process():
    global IMAGE_SERVER_PROCESS, IMAGE_SERVER_LOG_HANDLE, IMAGE_BASE_URL, IMAGE_PORT
    if IMAGE_SERVER_PROCESS is not None and IMAGE_SERVER_PROCESS.poll() is None:
        expected = Path(str(IMAGE_COMFY_DIR / 'main.py')).resolve()
        actual = Path(str(IMAGE_SERVER_PROCESS.args[1])).resolve()
        if actual != expected:
            raise RuntimeError('ComfyUI processの実引数が一致しないため停止しません。')
        IMAGE_SERVER_PROCESS.terminate()
        try:
            IMAGE_SERVER_PROCESS.wait(timeout=30)
        except subprocess.TimeoutExpired:
            IMAGE_SERVER_PROCESS.kill(); IMAGE_SERVER_PROCESS.wait(timeout=10)
    IMAGE_SERVER_PROCESS = None
    IMAGE_BASE_URL = None
    IMAGE_PORT = None
    if IMAGE_SERVER_LOG_HANDLE is not None:
        IMAGE_SERVER_LOG_HANDLE.close()
    IMAGE_SERVER_LOG_HANDLE = None

def _image_runtime_preflight():
    global IMAGE_VENV, IMAGE_COMFY_DIR, IMAGE_HOST, PRIVATE_PUBLIC_KEY_PATH, venv_python
    image_venv = Path(str(globals().get('IMAGE_VENV') or '/content/qwen_image_work/venv_fixed'))
    comfy_dir = Path(str(globals().get('IMAGE_COMFY_DIR') or '/content/qwen_image_work/ComfyUI'))
    IMAGE_VENV = image_venv
    IMAGE_COMFY_DIR = comfy_dir
    IMAGE_HOST = str(globals().get('IMAGE_HOST') or '127.0.0.1')
    key_value = globals().get('PRIVATE_PUBLIC_KEY_PATH')
    if not key_value:
        raise RuntimeError('画像準備セルを先に実行してください。')
    PRIVATE_PUBLIC_KEY_PATH = Path(str(key_value))
    candidate = globals().get('venv_python')
    if not candidate or not Path(str(candidate)).is_file():
        candidate = image_venv / 'bin' / 'python'
        if not candidate.is_file():
            candidate = image_venv / 'Scripts' / 'python.exe'
    gguf_name = str(globals().get('IMAGE_GGUF') or 'qwen-image-2.1-UC-Q8_0.gguf')
    te_name = Path(str(globals().get('IMAGE_TE') or 'qwen3vl_8b_int8_convrot.safetensors')).name
    vae_name = Path(str(globals().get('IMAGE_VAE') or 'qwen_image_2.1_vae_bf16.safetensors')).name
    required = [
        Path(str(candidate)),
        comfy_dir / 'main.py',
        comfy_dir / 'custom_nodes' / 'ComfyUI-GGUF',
        comfy_dir / 'custom_nodes' / 'qwen_private_node' / '__init__.py',
        PRIVATE_PUBLIC_KEY_PATH,
        comfy_dir / 'models' / 'unet' / gguf_name,
        comfy_dir / 'models' / 'text_encoders' / te_name,
        comfy_dir / 'models' / 'vae' / vae_name,
        Path(str(globals().get('PRIVATE_RESULT_DIR') or '')),
    ]
    if not str(globals().get('PRIVATE_ROUTE_BASE_URL') or '').strip():
        raise RuntimeError('画像準備セルを先に実行してください。')
    if not str(globals().get('PRIVATE_ROUTE_TOKEN') or '').strip():
        raise RuntimeError('画像準備セルを先に実行してください。')
    if any(not path.exists() for path in required):
        raise RuntimeError('画像準備セルを先に実行してください。')
    venv_python = Path(str(candidate))
    return venv_python

def start_image_backend():
    global IMAGE_PORT, IMAGE_BASE_URL, IMAGE_SERVER_PROCESS, IMAGE_SERVER_LOG_HANDLE
    if IMAGE_SERVER_PROCESS is not None and IMAGE_SERVER_PROCESS.poll() is None:
        if not IMAGE_BASE_URL:
            raise RuntimeError('既存ComfyUI processにendpointがありません。停止せずに起動を中止します。')
        if not _matches_comfy_process(IMAGE_SERVER_PROCESS):
            raise RuntimeError('既存processの実引数がComfyUIと一致しないため再利用しません。')
        try:
            if requests.get(f'{IMAGE_BASE_URL}/system_stats', timeout=3).status_code == 200:
                validate_image_object_info(IMAGE_BASE_URL)
                print('既存の検証済みComfyUIを再利用:', IMAGE_BASE_URL)
                return IMAGE_BASE_URL
        except Exception as exc:
            raise RuntimeError('既存ComfyUI processのhealthを確認できません。停止せずに起動を中止します。') from exc
    _image_runtime_preflight()
    if not _huihui_health():
        raise RuntimeError('Huihui serverのhealth/aliasが確認できません。画像backendを起動しません。')
    free_gib = _gpu_free_gib()
    if free_gib < IMAGE_MIN_FREE_VRAM_GIB:
        raise RuntimeError(f'画像backend用VRAM余裕不足: 空き{free_gib:.1f}GiB < {IMAGE_MIN_FREE_VRAM_GIB}GiB。Huihuiは稼働したままです。')
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as probe:
        probe.bind((IMAGE_HOST, 0)); IMAGE_PORT = probe.getsockname()[1]
    IMAGE_SERVER_LOG_HANDLE = open('/content/comfyui.log', 'a', encoding='utf-8')
    cmd = [str(venv_python), str(IMAGE_COMFY_DIR / 'main.py'), '--listen', IMAGE_HOST, '--port', str(IMAGE_PORT), '--disable-auto-launch', '--lowvram', '--reserve-vram', '2', '--cache-none', '--preview-method', 'none', '--disable-metadata']
    IMAGE_SERVER_PROCESS = subprocess.Popen(cmd, cwd=IMAGE_COMFY_DIR, stdout=IMAGE_SERVER_LOG_HANDLE, stderr=subprocess.STDOUT, text=True)
    IMAGE_BASE_URL = f'http://{IMAGE_HOST}:{IMAGE_PORT}'
    deadline = time.monotonic() + 180
    while time.monotonic() < deadline:
        if IMAGE_SERVER_PROCESS.poll() is not None:
            try:
                log_tail = Path('/content/comfyui.log').read_text(encoding='utf-8', errors='replace').splitlines()[-40:]
                safe_tail = [line for line in log_tail if not re.search(r'(?i)(token|secret|password|authorization)', line)]
                print('ComfyUI起動ログ（秘密情報らしい行は除外）:\n' + '\n'.join(safe_tail)[-4000:])
            except Exception:
                pass
            _clear_image_process()
            raise RuntimeError('ComfyUIがヘルスチェック前に終了しました。')
        try:
            if requests.get(f'{IMAGE_BASE_URL}/system_stats', timeout=3).status_code == 200:
                validate_image_object_info(IMAGE_BASE_URL)
                print('ComfyUI起動OK:', IMAGE_BASE_URL)
                return IMAGE_BASE_URL
        except requests.RequestException:
            pass
        except Exception:
            _clear_image_process()
            raise
        time.sleep(2)
    _clear_image_process()
    raise TimeoutError('ComfyUIが180秒以内に起動しませんでした')

print('準備完了後に start_image_backend() を実行してください。Huihuiを稼働させたままComfyUIを起動します。')
```

```python
try:
    from google.colab import output as colab_output
    colab_output.enable_custom_widget_manager()
except ImportError:
    pass

import io, json, ipywidgets as widgets
from IPython.display import clear_output, display

def create_image_ui():
    prompt = widgets.Textarea(placeholder='画像生成または編集の指示', layout=widgets.Layout(width='100%', height='90px'))
    upload = widgets.FileUpload(accept='image/*', multiple=False)
    generate = widgets.Button(description='画像を生成', button_style='primary')
    edit = widgets.Button(description='画像を編集')
    download = widgets.Button(description='暗号化ファイルをPCへ保存')
    ack = widgets.Button(description='PCで復号・保存を確認して削除')
    ack_check = widgets.Checkbox(description='PCで復号画像を保存済み', value=False)
    back = widgets.Button(description='チャットに戻る')
    pending = widgets.Dropdown(options=(), description='暗号化結果')
    output = widgets.Output(layout={'border': '1px solid #ddd'})
    state = {'busy': False, 'last_result': None, 'last_ref': None}
    global IMAGE_LAST_OUTPUT
    IMAGE_LAST_OUTPUT = None

    def auth_headers():
        return {'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}

    def refresh_pending():
        try:
            entries = json.loads((PRIVATE_RESULT_DIR / 'manifest.json').read_text(encoding='utf-8'))
            pending.options = [(str(item['result_id']), str(item['result_id'])) for item in entries if isinstance(item, dict) and item.get('result_id')]
        except Exception:
            pending.options = ()

    def selected_metadata():
        result_id = pending.value or (state['last_result'] or {}).get('opaque_id')
        if not result_id:
            return None
        response = requests.get(f'{PRIVATE_ROUTE_BASE_URL}/private/result/{result_id}/metadata', headers=auth_headers(), timeout=30)
        response.raise_for_status()
        metadata = response.json()
        # Normalize route metadata to the UI/node schema.
        metadata['opaque_id'] = metadata.get('opaque_id', metadata.get('result_id'))
        metadata['blob_name'] = metadata.get('blob_name', metadata.get('blob'))
        return metadata

    pending.observe(lambda _: setattr(ack_check, 'value', False), names='value')

    def uploaded_ref():
        if not upload.value:
            return None
        item = next(iter(upload.value.values())) if isinstance(upload.value, dict) else upload.value[0]
        content = item.get('content', b'')
        if not str(item.get('type', '')).startswith('image/') or len(content) > 25 * 1024 * 1024:
            raise ValueError('25MiB以下の画像を選択してください。')
        try:
            from PIL import Image
            normalized = io.BytesIO()
            Image.open(io.BytesIO(content)).convert('RGB').save(normalized, format='PNG')
            content = normalized.getvalue()
        except Exception as exc:
            raise ValueError('画像をPNGへ変換できません。') from exc
        response = requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference', headers=auth_headers(), data=content, timeout=60)
        response.raise_for_status()
        ref_id = response.json()['ref_id']
        try: upload.value = ()
        except Exception: pass
        state['last_ref'] = ref_id
        return ref_id

    def show_result(metadata):
        global IMAGE_LAST_OUTPUT
        state['last_result'] = metadata
        IMAGE_LAST_OUTPUT = metadata
        ack_check.value = False
        refresh_pending(); pending.value = metadata.get('opaque_id')
        with output:
            clear_output(wait=True)
            print('暗号化結果を準備しました。opaque_id:', metadata.get('opaque_id'))
            print('PCへ暗号化ファイルを保存し、ローカルviewerで復号してください。')

    def run(graph):
        if state['busy']:
            return
        state['busy'] = True
        generate.disabled = True; edit.disabled = True; download.disabled = True; ack.disabled = True; back.disabled = True
        try:
            with output: clear_output(wait=True); print('生成中…')
            def submit_and_wait():
                global GPU_INFERENCE_BLOCKED
                submitted = True
                try:
                    prompt_id = queue_image_graph(graph, base_url=IMAGE_BASE_URL)
                    return wait_image_result(prompt_id, base_url=IMAGE_BASE_URL)
                except Exception:
                    if submitted:
                        try: requests.post(f'{IMAGE_BASE_URL}/interrupt', timeout=10)
                        except Exception: pass
                        GPU_INFERENCE_BLOCKED = True
                    raise
            metadata, _ = run_gpu_serialized(submit_and_wait)
            show_result(metadata)
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass
            state['busy'] = False
            generate.disabled = False; edit.disabled = False; download.disabled = False; ack.disabled = False; back.disabled = False

    def on_generate(_):
        try:
            text = prompt.value.strip()
            if not text:
                with output: print('プロンプトを入力してください。')
                return
            run(build_t2i_graph(text))
        except Exception as exc:
            with output: print(f'画像生成エラー: {type(exc).__name__}')
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass

    def on_edit(_):
        try:
            text = prompt.value.strip()
            if not text:
                with output: print('編集指示を入力してください。')
                return
            ref_id = uploaded_ref() or state['last_ref']
            if not ref_id:
                raise ValueError('編集する画像をアップロードしてください。')
            run(build_i2i_graph(ref_id, text))
        except Exception as exc:
            with output: print(f'画像編集エラー: {type(exc).__name__}')
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass

    def on_download(_):
        try:
            metadata = selected_metadata()
            if not metadata: raise ValueError('保存する結果がありません。')
            from google.colab import files
            files.download(str(PRIVATE_RESULT_DIR / metadata['blob_name']))
            with output: print('暗号化ファイルの保存を開始しました。復号確認前に削除しません。')
        except Exception as exc:
            with output: print(f'暗号化ファイル保存エラー: {type(exc).__name__}')

    def on_ack(_):
        try:
            metadata = selected_metadata()
            if not metadata or not ack_check.value:
                with output: print('PCで復号画像を保存した後、確認欄を選択してください。')
                return
            response = requests.post(f"{PRIVATE_ROUTE_BASE_URL}/private/result/{metadata['opaque_id']}/ack", headers=auth_headers(), json={'recovered': True}, timeout=30)
            response.raise_for_status()
            state['last_result'] = None; IMAGE_LAST_OUTPUT = None; ack_check.value = False
            refresh_pending(); pending.value = None
            with output: clear_output(wait=True); print('回収確認済みの暗号化結果を削除しました。')
        except Exception as exc:
            with output: print(f'削除エラー: {type(exc).__name__}')

    def on_back(_):
        prompt.value = ''
        try: upload.value = ()
        except Exception: pass
        try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers=auth_headers(), json={}, timeout=10)
        except requests.RequestException: pass
        restore_huihui()

    generate.on_click(on_generate); edit.on_click(on_edit); download.on_click(on_download); ack.on_click(on_ack); back.on_click(on_back)
    refresh_pending()
    display(widgets.VBox([prompt, upload, pending, widgets.HBox([generate, edit, download, ack, back]), ack_check, output]))
    print('QWEN_PRIVATE_IMAGE_UI_READY', IMAGE_BASE_URL)

def stop_image_backend():
    if not GPU_INFERENCE_LOCK.acquire(blocking=False):
        raise RuntimeError('推論中はComfyUIを停止できません。推論完了後に再試行してください。')
    try:
        _clear_image_process()
    finally:
        GPU_INFERENCE_LOCK.release()
    try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers={'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}, json={}, timeout=10)
    except Exception: pass
    print('ComfyUIのみ停止しました。Huihuiは稼働したままです。')

def _huihui_health():
    base = str(globals().get('BASE_URL', '')).rstrip('/'); alias = str(globals().get('MODEL_ALIAS', ''))
    if not base or not alias: return False
    try:
        if requests.get(f'{base}/health', timeout=3).status_code != 200: return False
        ids = [str(x.get('id')) for x in requests.get(f'{base}/v1/models', timeout=3).json().get('data', [])]
        return alias in ids
    except requests.RequestException: return False

def restore_huihui():
    if GPU_INFERENCE_LOCK.locked():
        raise RuntimeError('推論中は画面を切り替えられません。完了後に再試行してください。')
    if not _huihui_health():
        raise RuntimeError('Huihui serverのhealth/aliasが確認できません。既存processは停止しません。')
    try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers={'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}, json={}, timeout=10)
    except Exception: pass
    chat_ui = create_chat_ui(BASE_URL, MODEL_ALIAS, initial_history=chat_history)
    display(widgets.VBox(list(chat_ui[:3]) + [chat_ui[3]]))
    print('HUIHUI_CHAT_REUSED', BASE_URL, MODEL_ALIAS, 'ComfyUI:', IMAGE_BASE_URL or '停止中')

start_image_button = widgets.Button(description='画像モードを表示', button_style='warning')
stop_image_button = widgets.Button(description='画像backend停止')
restore_huihui_button = widgets.Button(description='チャットに戻る')
def _start_image(_):
    try:
        start_image_backend()
        create_image_ui()
    except Exception as exc:
        print(f'画像backend開始エラー: {type(exc).__name__}: {exc}')
def _restore_huihui(_):
    try: restore_huihui()
    except Exception as exc: print(f'Huihui復帰エラー: {type(exc).__name__}')
start_image_button.on_click(_start_image); stop_image_button.on_click(lambda _: stop_image_backend()); restore_huihui_button.on_click(_restore_huihui)
stop_all_button = widgets.Button(description='両バックエンドを停止', button_style='danger')
def stop_all_backends(_):
    global GPU_INFERENCE_BLOCKED, CONTINUE_BRIDGE
    bridge = globals().get('CONTINUE_BRIDGE')
    if bridge is not None:
        try: bridge.stop_accepting()
        except Exception: pass
    if not GPU_INFERENCE_LOCK.acquire(blocking=False):
        print('推論中はサーバーを停止できません。接続を閉じました。完了後にもう一度停止してください。')
        return
    try:
        GPU_INFERENCE_BLOCKED = True
        _clear_image_process()
        try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers={'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}, json={}, timeout=10)
        except Exception: pass
        stop_server()
        if IMAGE_SERVER_PROCESS is not None or server_process is not None:
            raise RuntimeError('両backendの停止を確認できません。推論ブロックを解除しません。')
        if bridge is not None:
            try: bridge.session.close()
            except Exception: pass
            CONTINUE_BRIDGE = None
        GPU_INFERENCE_BLOCKED = False
        print('このノートブックが起動した両バックエンドを停止しました。')
    except Exception as exc:
        print(f'停止エラー: {type(exc).__name__}: {exc}')
    finally:
        GPU_INFERENCE_LOCK.release()
stop_all_button.on_click(stop_all_backends)
display(widgets.HBox([start_image_button, stop_image_button, restore_huihui_button, stop_all_button]))
try:
    start_image_backend()
    create_image_ui()
    if not _huihui_health():
        raise RuntimeError('Huihui health/alias確認に失敗しました。')
    image_health = requests.get(f'{IMAGE_BASE_URL}/system_stats', timeout=10)
    image_health.raise_for_status()
    print('DUAL_BACKENDS_READY', 'Huihui:', BASE_URL, MODEL_ALIAS, 'PID', server_process.pid, 'ComfyUI:', IMAGE_BASE_URL, 'PID', IMAGE_SERVER_PROCESS.pid)
except Exception as exc:
    print(f'画像backendを準備できませんでした。Huihui processは稼働したままです: {type(exc).__name__}: {exc}')
```

```python
# Optional VS Code connection: helper is fetched from this repository and SHA256-verified. No tunnel starts on Run All.
import hashlib, importlib.util, shutil, subprocess
from pathlib import Path

CONTINUE_BRIDGE_SOURCE = 'https://raw.githubusercontent.com/jialanfudao-stack/git-practice/main/continue_colab_bridge.py'
CONTINUE_BRIDGE_SHA256 = 'ec716c5757183c037b289553b71dd8a75c71c5213ee080bf12035c59f10db5de'
CONTINUE_BRIDGE_FILE = Path('/content/continue_colab_bridge.py')
if CONTINUE_BRIDGE_FILE.is_file() and hashlib.sha256(CONTINUE_BRIDGE_FILE.read_bytes()).hexdigest() == CONTINUE_BRIDGE_SHA256:
    helper_bytes = CONTINUE_BRIDGE_FILE.read_bytes()
    print('SHA256-verified Continue helper reused')
else:
    response = requests.get(CONTINUE_BRIDGE_SOURCE, timeout=30)
    response.raise_for_status()
    helper_bytes = response.content
    if hashlib.sha256(helper_bytes).hexdigest() != CONTINUE_BRIDGE_SHA256:
        raise RuntimeError('VS Code bridge helper SHA256 verification failed. Sync the notebook and helper first.')
    CONTINUE_BRIDGE_FILE.write_bytes(helper_bytes)
spec = importlib.util.spec_from_file_location('continue_colab_bridge', CONTINUE_BRIDGE_FILE)
continue_bridge_module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(continue_bridge_module)
ColabContinueBridge = continue_bridge_module.ColabContinueBridge
CONTINUE_BRIDGE = globals().get('CONTINUE_BRIDGE')

def _cloudflared_binary():
    binary = shutil.which('cloudflared')
    if binary:
        return binary
    package_path = Path('/content/cloudflared-official.deb')
    package_response = requests.get(
        'https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb',
        timeout=120,
    )
    package_response.raise_for_status()
    package_path.write_bytes(package_response.content)
    install = subprocess.run(['dpkg', '-i', str(package_path)], capture_output=True, text=True)
    if install.returncode:
        raise RuntimeError('Cloudflare公式cloudflared packageをinstallできません。Colab runtime権限と依存関係を確認してください。')
    binary = shutil.which('cloudflared')
    if not binary:
        raise RuntimeError('cloudflared install後にbinaryが見つかりません。')
    return binary

def _start_continue_connection(_):
    global CONTINUE_BRIDGE
    try:
        if not _huihui_health():
            raise RuntimeError('文章Qwenのhealth/aliasを確認できません。')
        if IMAGE_SERVER_PROCESS is None or IMAGE_SERVER_PROCESS.poll() is not None:
            raise RuntimeError('ComfyUIが起動していません。両backendの準備後に接続してください。')
        if CONTINUE_BRIDGE is None or not CONTINUE_BRIDGE.accepting:
            CONTINUE_BRIDGE = ColabContinueBridge(
                BASE_URL,
                MODEL_ALIAS,
                GPU_INFERENCE_LOCK,
                lambda: GPU_INFERENCE_BLOCKED,
                lambda: globals().__setitem__('GPU_INFERENCE_BLOCKED', True),
            )
        binary = _cloudflared_binary()
        api_base = CONTINUE_BRIDGE.start_tunnel(binary)
        CONTINUE_BRIDGE.write_config('/content/colab_continue_connection.json')
        print('VS Code接続準備完了:', api_base, 'model:', MODEL_ALIAS, '（認証キーは設定JSONに保存しました）')
    except Exception as exc:
        if CONTINUE_BRIDGE is not None and not CONTINUE_BRIDGE.api_base:
            try: CONTINUE_BRIDGE.stop()
            except Exception: pass
            CONTINUE_BRIDGE = None
        print('VS Code接続エラー:', type(exc).__name__, str(exc))

def _download_continue_config(_):
    if CONTINUE_BRIDGE is None or not CONTINUE_BRIDGE.api_base:
        print('先に「VS Code接続を開始」を押してください。')
        return
    try:
        from google.colab import files
        files.download(CONTINUE_BRIDGE.write_config('/content/colab_continue_connection.json'))
    except Exception as exc:
        print('設定JSONをダウンロードできません:', type(exc).__name__)

def _stop_continue_connection(_):
    global CONTINUE_BRIDGE
    bridge = CONTINUE_BRIDGE
    if bridge is None:
        print('VS Code接続は起動していません。')
        return
    try:
        bridge.stop_accepting()
        if GPU_INFERENCE_LOCK.acquire(blocking=False):
            try: bridge.session.close()
            finally: GPU_INFERENCE_LOCK.release()
        CONTINUE_BRIDGE = None
        print('VS Codeの公開接続を停止しました。')
    except Exception as exc:
        print('VS Code接続を停止できません:', type(exc).__name__)

continue_start_button = widgets.Button(description='VS Code接続を開始', button_style='primary')
continue_download_button = widgets.Button(description='Continue設定JSONをPCへ保存')
continue_stop_button = widgets.Button(description='VS Code接続を停止')
continue_start_button.on_click(_start_continue_connection)
continue_download_button.on_click(_download_continue_config)
continue_stop_button.on_click(_stop_continue_connection)
display(widgets.HBox([continue_start_button, continue_download_button, continue_stop_button]))
```

## 通常再実行（初回準備済み・両backend稼働中）

同じColab runtimeでHuihuiとComfyUIが稼働中なら、次の1セルだけを実行してください。既存process、推論lock、blocked状態、会話履歴を検査してからUI/function definitionsを更新します。model download、pip、build、秘密鍵setup、推論、backend起動は行いません。状態不明・不一致なら自動で初回準備へ進まず、説明を出して停止します。

ランタイム再起動後、初回実行前、backend停止中、推論中、またはlockがblockedのときはこのセルを使わず、下の「初回準備」手順から実行してください。

```python
# @title 通常再実行（初回準備済み・両backend稼働中のみ）
# This cell only refreshes UI/function definitions from the pinned, verified notebook.
import hashlib, json, requests
from pathlib import Path

NOTEBOOK_REVISION = '6c3b6a0450df87f7494718a3f6e693e6adde8c9f'
NOTEBOOK_SHA256 = '702fe767ad833aa66b509c687159801508603786b3e456e21a5068d8932a747d'
NOTEBOOK_URL = f'https://raw.githubusercontent.com/jialanfudao-stack/git-practice/{NOTEBOOK_REVISION}/Untitled0.ipynb'


def _rerun_refuse(reason):
    raise RuntimeError(f'通常再実行を中止しました: {reason}。初回準備が必要か、runtime状態を確認してください。')


_required = ('GPU_INFERENCE_LOCK', 'GPU_INFERENCE_BLOCKED', 'chat_history', 'BASE_URL', 'MODEL_ALIAS',
             'SERVER_BIN', 'model_path', 'server_process', 'IMAGE_COMFY_DIR', 'IMAGE_BASE_URL',
             'IMAGE_SERVER_PROCESS', 'requests', 'widgets')
_missing = [name for name in _required if name not in globals() or globals()[name] is None]
if _missing:
    _rerun_refuse('必要な準備済み状態がありません (' + ', '.join(_missing) + ')')
if GPU_INFERENCE_BLOCKED:
    _rerun_refuse('GPU共有guardがblocked状態です')
if not isinstance(chat_history, list):
    _rerun_refuse('chat_historyの形式を確認できません')
_REFRESH_LOCK = GPU_INFERENCE_LOCK
if not _REFRESH_LOCK.acquire(blocking=False):
    _rerun_refuse('推論中のためGPU共有lockが使用中です')
try:
    if GPU_INFERENCE_BLOCKED:
        _rerun_refuse('GPU共有guardがblocked状態です')


    def _live(proc):
        return proc is not None and proc.poll() is None


    def _argv(proc):
        value = proc.args
        return list(value) if isinstance(value, (list, tuple)) else str(value).split()


    if not _live(server_process) or not _live(IMAGE_SERVER_PROCESS):
        _rerun_refuse('HuihuiまたはComfyUIの既存processが稼働していません')
    try:
        import psutil
        _server_argv = psutil.Process(server_process.pid).cmdline()
        _image_argv = psutil.Process(IMAGE_SERVER_PROCESS.pid).cmdline()
        _server_exe = Path(_server_argv[0]).resolve()
        _server_model = Path(_server_argv[_server_argv.index('-m') + 1]).resolve()
        if (_server_exe != Path(str(SERVER_BIN)).resolve()
                or _server_model != Path(str(model_path)).resolve()
                or _server_argv[_server_argv.index('--alias') + 1] != MODEL_ALIAS):
            _rerun_refuse('Huihui processの実行ファイル/model/aliasが一致しません')
        if (len(_image_argv) < 2
                or Path(_image_argv[1]).resolve() != Path(str(IMAGE_COMFY_DIR) + '/main.py').resolve()):
            _rerun_refuse('ComfyUI processの実体を確認できません')
    except Exception as exc:
        if isinstance(exc, RuntimeError):
            raise
        _rerun_refuse('既存processの実体を検証できません (' + type(exc).__name__ + ')')

    try:
        _health = requests.get(str(BASE_URL).rstrip('/') + '/health', timeout=3)
        _models = requests.get(str(BASE_URL).rstrip('/') + '/v1/models', timeout=3)
        _model_ids = [str(item.get('id')) for item in _models.json().get('data', [])]
        _image_health = requests.get(str(IMAGE_BASE_URL).rstrip('/') + '/system_stats', timeout=5)
        if _health.status_code != 200 or _models.status_code != 200 or MODEL_ALIAS not in _model_ids:
            _rerun_refuse('Huihui health/aliasを確認できません')
        _image_health.raise_for_status()
    except Exception as exc:
        if isinstance(exc, RuntimeError):
            raise
        _rerun_refuse('backend healthを確認できません (' + type(exc).__name__ + ')')

    _preserved = {
        'GPU_INFERENCE_LOCK': GPU_INFERENCE_LOCK,
        'GPU_INFERENCE_BLOCKED': GPU_INFERENCE_BLOCKED,
        'chat_history': chat_history,
        'history_values': list(chat_history),
        'server_process': server_process,
        'IMAGE_SERVER_PROCESS': IMAGE_SERVER_PROCESS,
        'CONTINUE_BRIDGE': globals().get('CONTINUE_BRIDGE'),
    }
    if _preserved['CONTINUE_BRIDGE'] is not None:
        _bridge = _preserved['CONTINUE_BRIDGE']
        if getattr(_bridge, 'gpu_lock', GPU_INFERENCE_LOCK) is not GPU_INFERENCE_LOCK:
            _rerun_refuse('既存Continue bridgeが共有GPU lockを使っていません')
        _tunnel = getattr(_bridge, 'tunnel_process', None)
        if getattr(_bridge, 'api_base', None) and not _live(_tunnel):
            _rerun_refuse('既存Continue tunnelの状態を確認できません')

    _response = requests.get(NOTEBOOK_URL, timeout=30)
    _response.raise_for_status()
    _notebook_bytes = _response.content
    if hashlib.sha256(_notebook_bytes).hexdigest() != NOTEBOOK_SHA256:
        _rerun_refuse('固定revisionのNotebook SHA256が一致しません。同期済みNotebookを確認してください')
    _source_nb = json.loads(_notebook_bytes)

    # Only these definitions/UI cells are used. Preparation, model downloads, pip/build,
    # key setup, smoke inference, and backend startup cells are intentionally excluded.
    for _index in (13, 15, 19, 20, 21, 34):
        if _index >= len(_source_nb.get('cells', [])) or _source_nb['cells'][_index].get('cell_type') != 'code':
            _rerun_refuse(f'軽量セル {_index} のNotebook構造が一致しません')
        _cell_source = ''.join(_source_nb['cells'][_index].get('source', []))
        if _index == 13:
            _history_reset = 'chat_history = []'
            if _cell_source.count(_history_reset) != 1:
                _rerun_refuse('推論関数セルの履歴初期化箇所を検証できません')
            _cell_source = _cell_source.replace(_history_reset, "chat_history = _preserved['chat_history']")
        if _index == 15:
            _old_call = 'create_chat_ui(BASE_URL, MODEL_ALIAS)'
            if _cell_source.count(_old_call) != 1:
                _rerun_refuse('chat UIの履歴保持箇所を検証できません')
            _cell_source = _cell_source.replace(_old_call, 'create_chat_ui(BASE_URL, MODEL_ALIAS, initial_history=chat_history)')
        if _index == 21:
            _auto_start = '\ntry:\n    start_image_backend()'
            if _cell_source.count(_auto_start) != 1:
                _rerun_refuse('画像backendの自動起動箇所を安全に除外できません')
            _cell_source = _cell_source.split(_auto_start, 1)[0]
        exec(compile(_cell_source, f'<verified-notebook-cell-{_index}>', 'exec'), globals())

    if (GPU_INFERENCE_LOCK is not _preserved['GPU_INFERENCE_LOCK']
            or GPU_INFERENCE_BLOCKED != _preserved['GPU_INFERENCE_BLOCKED']
            or chat_history is not _preserved['chat_history']
            or chat_history != _preserved['history_values']
            or server_process is not _preserved['server_process']
            or IMAGE_SERVER_PROCESS is not _preserved['IMAGE_SERVER_PROCESS']
            or globals().get('CONTINUE_BRIDGE') is not _preserved['CONTINUE_BRIDGE']):
        _rerun_refuse('refresh中に共有guard/history/processが予期せず変わりました')
    if GPU_INFERENCE_BLOCKED:
        _rerun_refuse('refresh後の共有guardがblocked状態です')
    print('通常再実行完了: 既存Huihui/ComfyUIと共有guardを維持し、UI/function definitionsだけ更新しました。')
    print('初回準備セル、model download、pip、build、key setup、backend起動は実行していません。')
finally:
    _REFRESH_LOCK.release()
```
