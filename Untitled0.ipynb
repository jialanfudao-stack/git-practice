```python
try:
    from google.colab import output as colab_output
    colab_output.enable_custom_widget_manager()
except ImportError:
    pass

import ipywidgets as widgets
from IPython.display import clear_output, display

def create_chat_ui(base_url, model_alias, initial_history=None):
    """Create a chat UI whose callbacks keep one verified server endpoint."""
    base_url = str(base_url).rstrip('/')
    model_alias = str(model_alias)
    history = list(initial_history or [])

    def check_server():
        health = requests.get(f'{base_url}/health', timeout=10)
        health.raise_for_status()
        models_response = requests.get(f'{base_url}/v1/models', timeout=10)
        models_response.raise_for_status()
        model_ids = [str(item.get('id')) for item in models_response.json().get('data', [])]
        if model_alias not in model_ids:
            raise RuntimeError(f'接続先のモデルaliasが一致しません: expected={model_alias}, actual={model_ids}')

    def trim_history(items):
        pairs = []
        pending = None
        for item in items:
            if item.get('role') == 'user':
                pending = item
            elif item.get('role') == 'assistant' and pending is not None:
                pairs.append((pending, item))
                pending = None
        kept = []
        total = 0
        for pair in reversed(pairs[-MAX_HISTORY_TURNS:]):
            chars = sum(len(str(item.get('content', ''))) for item in pair)
            if total + chars > MAX_HISTORY_CHARS:
                break
            kept.append(pair)
            total += chars
        return [item for pair in reversed(kept) for item in pair]

    def token_count(messages):
        response = requests.post(
            f'{base_url}/apply-template',
            json={'messages': messages},
            timeout=30,
        )
        response.raise_for_status()
        prompt = response.json().get('prompt')
        if not isinstance(prompt, str):
            raise RuntimeError('/apply-templateのpromptがありません。')
        response = requests.post(
            f'{base_url}/tokenize',
            json={'content': prompt, 'add_special': False},
            timeout=30,
        )
        response.raise_for_status()
        tokens = response.json().get('tokens')
        if not isinstance(tokens, list):
            raise RuntimeError('/tokenizeのtokensがありません。')
        return len(tokens)

    def generate(user_text, previous=None, max_tokens=MAX_NEW_TOKENS):
        user_text = user_text.strip()
        if not user_text:
            raise ValueError('質問を入力してください。')
        check_server()
        previous = trim_history(previous or [])
        while True:
            messages = [
                {'role': 'system', 'content': SYSTEM_PROMPT},
                *previous,
                {'role': 'user', 'content': user_text},
            ]
            prompt_tokens = token_count(messages)
            if prompt_tokens <= MAX_PROMPT_TOKENS:
                break
            if previous:
                previous = previous[2:]
                continue
            raise ValueError(
                f'入力が長すぎます（テンプレート後{prompt_tokens}トークン）。'
                f'質問を短くして、{MAX_PROMPT_TOKENS}トークン以内にしてください。'
            )
        payload = {
            'model': model_alias,
            'messages': messages,
            'max_tokens': max_tokens,
            'temperature': 0.7,
            'top_p': 0.8,
            'reasoning_effort': 'none',
            'chat_template_kwargs': {'enable_thinking': False},
        }
        response = requests.post(
            f'{base_url}/v1/chat/completions',
            json=payload,
            timeout=900,
        )
        response.raise_for_status()
        data = response.json()
        try:
            answer = data['choices'][0]['message']['content']
        except (KeyError, IndexError, TypeError) as exc:
            raise RuntimeError('chat completions応答形式が不正です。') from exc
        return str(answer).strip()

    prompt_box = widgets.Textarea(
        placeholder='質問を入力してください',
        layout=widgets.Layout(width='100%', height='90px'),
    )
    send_button = widgets.Button(description='送信', button_style='primary')
    reset_button = widgets.Button(description='会話をリセット')
    chat_output = widgets.Output(layout={'border': '1px solid #ddd'})

    def send(_):
        text = prompt_box.value.strip()
        if not text:
            return
        send_button.disabled = True
        try:
            answer = generate(text, previous=history)
            history[:] = trim_history(
                history + [
                    {'role': 'user', 'content': text},
                    {'role': 'assistant', 'content': answer},
                ]
            )
            chat_history[:] = history
            with chat_output:
                print(f'あなた: {text}')
                print(f'Huihui Qwen3.8 Q6_K_L: {answer}')
                print()
            prompt_box.value = ''
        except Exception as exc:
            with chat_output:
                print(f'エラー: {type(exc).__name__}')
        finally:
            prompt_box.value = ''
            send_button.disabled = False

    def reset(_):
        history.clear()
        chat_history.clear()
        prompt_box.value = ''
        with chat_output:
            clear_output(wait=True)
            print('会話をリセットしました。')

    check_server()
    send_button.on_click(send)
    reset_button.on_click(reset)
    return prompt_box, send_button, reset_button, chat_output

clear_output(wait=True)
prompt_box, send_button, reset_button, chat_output = create_chat_ui(BASE_URL, MODEL_ALIAS)
display(widgets.VBox([prompt_box, widgets.HBox([send_button, reset_button]), chat_output]))
print('HUIHUI_CHAT_UI_READY', BASE_URL, MODEL_ALIAS)
```


## Qwen-Image-2.1（T2I + 画像編集）

公式のworkflowはUI subgraphなので、ここではT2I/I2Iに必要なノードだけをAPI graphへ展開します。画像backend停止後は「Huihuiへ戻る」セルでチャットを再起動できます。

```python

# Image preparation: this cell does not stop the Huihui server.
import hashlib, json, os, shutil, subprocess, sys, time
from pathlib import Path
from huggingface_hub import HfApi, hf_hub_download

IMAGE_MIN_FREE_DISK_GIB = 45
IMAGE_WORK_DIR = Path('/content/qwen_image_work')
IMAGE_COMFY_DIR = IMAGE_WORK_DIR / 'ComfyUI'
IMAGE_VENV = IMAGE_WORK_DIR / 'venv_fixed'
IMAGE_MODELS = IMAGE_COMFY_DIR / 'models'
IMAGE_REVISION = '6b34e59458d3eb7ba6a6f86a116aed5253dc02c3'
IMAGE_REPO = 'abenzerps/Qwen-Image-2.1-Uncensored-GGUF'
IMAGE_GGUF = 'qwen-image-2.1-UC-Q8_0.gguf'
IMAGE_GGUF_BYTES = 7591557920
IMAGE_GGUF_SHA256 = 'cde456c72ea3ecebfc1be783300e972711d875e0c5f1bed33d42b66b156affa8'
COMFY_REPO = 'Comfy-Org/Qwen-Image-2.1'
COMFY_REVISION = 'cb504a4090723e43f17ad01cec0359490e2de613'
IMAGE_TE = 'text_encoders/qwen3vl_8b_int8_convrot.safetensors'
IMAGE_TE_BYTES = 9350798360
IMAGE_TE_SHA256 = '8bfd0f6e12abf2d2d697ecc888e5e90b0d6741d6708f05799f53afa560452e8f'
IMAGE_VAE = 'vae/qwen_image_2.1_vae_bf16.safetensors'
IMAGE_VAE_BYTES = 675509688
IMAGE_VAE_SHA256 = 'bb21f7473051e1ac368515dd3f2e15cd44d7a11748ee8823e1ddca3e4876b7c9'
COMFY_COMMIT = '651ca296a73cd21c12a57eb8741d52e40dc6528f'
GGUF_NODE_COMMIT = '373048b8403a7820620065210a691263d4da0a61'
GGUF_NODE_REPO = 'https://github.com/leejet/ComfyUI-GGUF'

free_gib = shutil.disk_usage('/content').free / (1024**3)
try:
    import psutil
    ram_available_gib = psutil.virtual_memory().available / (1024**3)
except Exception as exc:
    raise RuntimeError('RAM空き容量を確認できないため、画像モデルを準備しません。') from exc
if ram_available_gib < 8:
    raise RuntimeError(f'RAM空き容量不足: {ram_available_gib:.1f}GiB。Huihuiは停止していません。')
if free_gib < IMAGE_MIN_FREE_DISK_GIB:
    raise RuntimeError(f'/contentの空き容量が不足しています: {free_gib:.1f}GiB < {IMAGE_MIN_FREE_DISK_GIB}GiB。Huihuiは停止していません。')
if 'A100' not in str(gpu_name).upper() or vram_gib < 35:
    raise RuntimeError(f'画像モデル用GPU/VRAM不足: {gpu_name}, {vram_gib:.1f}GiB。Huihuiは停止していません。')
print(f'画像準備前チェックOK: {gpu_name}, {vram_gib:.1f}GiB, RAM空き{ram_available_gib:.1f}GiB, 空き{free_gib:.1f}GiB')

def sha256_file(path):
    h = hashlib.sha256()
    with Path(path).open('rb') as f:
        for block in iter(lambda: f.read(16 * 1024 * 1024), b''):
            h.update(block)
    return h.hexdigest()

def hf_verified(repo_id, filename, revision, target, expected_bytes, expected_sha, download_root):
    target = Path(target)
    target.parent.mkdir(parents=True, exist_ok=True)
    download_root = Path(download_root)
    download_root.mkdir(parents=True, exist_ok=True)
    if target.exists() and target.stat().st_size == expected_bytes and sha256_file(target) == expected_sha:
        print('既存ファイルのSHA256検証OK:', target)
        return target
    downloaded = Path(hf_hub_download(repo_id=repo_id, filename=filename, revision=revision, local_dir=download_root))
    if downloaded != target:
        target.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(downloaded, target)
    if target.stat().st_size != expected_bytes:
        raise RuntimeError(f'サイズ不一致: {target} ({target.stat().st_size} != {expected_bytes})')
    actual = sha256_file(target)
    if actual != expected_sha:
        raise RuntimeError(f'SHA256不一致: {target}\nexpected={expected_sha}\nactual={actual}')
    print('取得・SHA256検証OK:', target)
    return target

IMAGE_WORK_DIR.mkdir(parents=True, exist_ok=True)

if not (IMAGE_COMFY_DIR / '.git').exists():
    subprocess.run(['git', 'clone', 'https://github.com/Comfy-Org/ComfyUI', str(IMAGE_COMFY_DIR)], check=True)
subprocess.run(['git', '-C', str(IMAGE_COMFY_DIR), 'fetch', '--depth=1', 'origin', COMFY_COMMIT], check=True)
subprocess.run(['git', '-C', str(IMAGE_COMFY_DIR), 'checkout', '--detach', COMFY_COMMIT], check=True)
custom_dir = IMAGE_COMFY_DIR / 'custom_nodes' / 'ComfyUI-GGUF'
if not (custom_dir / '.git').exists():
    custom_dir.parent.mkdir(parents=True, exist_ok=True)
    subprocess.run(['git', 'clone', GGUF_NODE_REPO, str(custom_dir)], check=True)
subprocess.run(['git', '-C', str(custom_dir), 'fetch', '--depth=1', 'origin', GGUF_NODE_COMMIT], check=True)
subprocess.run(['git', '-C', str(custom_dir), 'checkout', '--detach', GGUF_NODE_COMMIT], check=True)

IMAGE_MODELS.mkdir(parents=True, exist_ok=True)
hf_verified(IMAGE_REPO, IMAGE_GGUF, IMAGE_REVISION, IMAGE_MODELS / 'unet' / IMAGE_GGUF, IMAGE_GGUF_BYTES, IMAGE_GGUF_SHA256, download_root=IMAGE_MODELS / 'unet')
hf_verified(COMFY_REPO, IMAGE_TE, COMFY_REVISION, IMAGE_MODELS / 'text_encoders' / Path(IMAGE_TE).name, IMAGE_TE_BYTES, IMAGE_TE_SHA256, download_root=IMAGE_MODELS)
hf_verified(COMFY_REPO, IMAGE_VAE, COMFY_REVISION, IMAGE_MODELS / 'vae' / Path(IMAGE_VAE).name, IMAGE_VAE_BYTES, IMAGE_VAE_SHA256, download_root=IMAGE_MODELS)

def _venv_python(path):
    path = Path(path)
    py = path / 'bin' / 'python'
    return py if py.exists() else path / 'Scripts' / 'python.exe'

def _venv_probe(path):
    path = Path(path)
    py = _venv_python(path)
    if not py.exists():
        return False, 'python executable missing'
    probe = subprocess.run([str(py), '-c', 'import sys, pip; print(sys.prefix); print(sys.base_prefix); print(pip.__version__)'], capture_output=True, text=True)
    lines = probe.stdout.strip().splitlines()
    expected = str(path.resolve())
    ok = probe.returncode == 0 and len(lines) >= 3 and str(Path(lines[0]).resolve()) == expected
    detail = probe.stdout.strip() or probe.stderr.strip()
    return ok, detail

venv_ok, venv_detail = _venv_probe(IMAGE_VENV)
if not venv_ok:
    subprocess.run([sys.executable, '-m', 'venv', '--without-pip', '--system-site-packages', str(IMAGE_VENV)], check=True)
    venv_ok, venv_detail = _venv_probe(IMAGE_VENV)
    if not venv_ok:
        raise RuntimeError('独立venv検証失敗: ' + str(venv_detail))
venv_python = _venv_python(IMAGE_VENV)
print('IMAGE_VENV_READY', IMAGE_VENV, venv_detail.replace('\n', ' | '))
subprocess.run([str(venv_python), '-m', 'pip', 'install', '--disable-pip-version-check', '-q', '-r', str(IMAGE_COMFY_DIR / 'requirements.txt')], check=True)
subprocess.run([str(venv_python), '-m', 'pip', 'install', '--disable-pip-version-check', '-q', 'gguf>=0.13.0'], check=True)
print('ComfyUI/ComfyUI-GGUF固定commitと独立venvの準備OK')
print('Q8_0/INT8 text encoder/VAEの取得とSHA256検証OK。ここまでHuihuiは稼働中です。')
```

```python

# Explicit minimal API graphs. Validate node names and inputs against /object_info after ComfyUI starts.
import io, json, threading, uuid, requests
from IPython.display import display

IMAGE_HOST = '127.0.0.1'
IMAGE_MODEL_ALIAS = 'qwen-image-2.1-uc-q8'
IMAGE_PORT = globals().get('IMAGE_PORT')
IMAGE_BASE_URL = globals().get('IMAGE_BASE_URL')
IMAGE_SERVER_PROCESS = globals().get('IMAGE_SERVER_PROCESS')
IMAGE_SERVER_LOG_HANDLE = globals().get('IMAGE_SERVER_LOG_HANDLE')
IMAGE_LAST_OUTPUT = globals().get('IMAGE_LAST_OUTPUT')

def validate_image_object_info(base_url):
    info = requests.get(f'{base_url}/object_info', timeout=30).json()
    required = {'UnetLoaderGGUF', 'CLIPLoader', 'VAELoader', 'TextEncodeQwenImage21', 'QwenImage21Cache', 'EmptyLatentImage', 'KSampler', 'VAEDecode', 'QwenEncryptedSaveImage', 'QwenMemoryReference'}
    missing = sorted(required - set(info))
    if missing:
        raise RuntimeError(f'ComfyUI object_infoに必要ノードがありません: {missing}')
    for cls, names in {
        'TextEncodeQwenImage21': {'clip', 'prompt', 'negative_prompt', 'resolution'},
        'QwenImage21Cache': {'model', 'device', 'dtype'},
    }.items():
        inputs = info[cls].get('input', {})
        names_present = set(inputs.get('required', {})) | set(inputs.get('optional', {}))
        if not names <= names_present:
            raise RuntimeError(f'{cls}の入力schemaが想定外です: {sorted(names_present)}')
    print('object_info schema OK:', ', '.join(sorted(required)))

def build_t2i_graph(prompt, width=1024, height=1024, seed=20261001, steps=25):
    return {
        '1': {'class_type': 'UnetLoaderGGUF', 'inputs': {'unet_name': IMAGE_GGUF}},
        '2': {'class_type': 'CLIPLoader', 'inputs': {'clip_name': Path(IMAGE_TE).name, 'type': 'qwen_image'}},
        '3': {'class_type': 'VAELoader', 'inputs': {'vae_name': Path(IMAGE_VAE).name}},
        '4': {'class_type': 'TextEncodeQwenImage21', 'inputs': {'clip': ['2', 0], 'prompt': prompt, 'negative_prompt': '', 'resolution': max(width, height)}},
        '5': {'class_type': 'QwenImage21Cache', 'inputs': {'model': ['1', 0], 'device': 'auto', 'dtype': 'default'}},
        '6': {'class_type': 'EmptyLatentImage', 'inputs': {'width': width, 'height': height, 'batch_size': 1}},
        '7': {'class_type': 'KSampler', 'inputs': {'model': ['5', 0], 'positive': ['4', 0], 'negative': ['4', 1], 'latent_image': ['6', 0], 'seed': seed, 'steps': steps, 'cfg': 1.0, 'sampler_name': 'euler', 'scheduler': 'simple', 'denoise': 1.0}},
        '8': {'class_type': 'VAEDecode', 'inputs': {'samples': ['7', 0], 'vae': ['3', 0]}},
        '9': {'class_type': 'QwenEncryptedSaveImage', 'inputs': {'image': ['8', 0], 'public_key_path': str(PRIVATE_PUBLIC_KEY_PATH), 'result_id': f'qwen-image-{uuid.uuid4().hex}'}},
    }

def build_i2i_graph(ref_id, prompt, width=1024, height=1024, seed=20261001, steps=25):
    graph = build_t2i_graph(prompt, width, height, seed, steps)
    graph['10'] = {'class_type': 'QwenMemoryReference', 'inputs': {'ref_id': ref_id}}
    graph['4']['inputs'].update({'images.image_1': ['10', 0], 'vae': ['3', 0]})
    graph['7']['inputs']['latent_image'] = ['4', 2]
    return graph

def queue_image_graph(graph, base_url=None):
    base_url = base_url or IMAGE_BASE_URL
    if not base_url:
        raise RuntimeError('画像backendが起動していません。')
    payload = {'prompt': graph, 'client_id': str(uuid.uuid4())}
    response = requests.post(f'{base_url}/prompt', json=payload, timeout=30)
    response.raise_for_status()
    return response.json()['prompt_id']

def clear_comfy_history(base_url, prompt_id):
    try:
        response = requests.post(
            f'{base_url}/history',
            json={'delete': [str(prompt_id)]},
            timeout=10,
        )
        response.raise_for_status()
        return True
    except requests.RequestException:
        return False

def wait_image_result(prompt_id, base_url=None, timeout=900):
    base_url = base_url or IMAGE_BASE_URL
    if not base_url:
        raise RuntimeError('画像backendが起動していません。')
    deadline = time.monotonic() + timeout
    result = None
    try:
        while time.monotonic() < deadline:
            response = requests.get(f'{base_url}/history/{prompt_id}', timeout=30)
            response.raise_for_status()
            history = response.json().get(prompt_id)
            if history and history.get('status', {}).get('status_str') == 'error':
                raise RuntimeError('画像生成に失敗しました。詳細な履歴は表示しません。')
            if history and history.get('outputs'):
                for node in history['outputs'].values():
                    for item in node.get('private_result', []):
                        result = item
                        break
                    if result is not None:
                        break
            if result is not None:
                break
            time.sleep(2)
        if result is None:
            raise TimeoutError('画像生成が制限時間内に完了しませんでした。')
        return result, None
    finally:
        if not clear_comfy_history(base_url, prompt_id):
            raise RuntimeError('生成履歴の消去に失敗しました。結果は表示していません。')
```

```python
# After the preparation cell reports success, stop only our Huihui process and start ComfyUI on a dynamic localhost port.
HUIHUI_RESTORE_CMD = list(globals().get('HUIHUI_RESTORE_CMD') or globals().get('server_cmd', []))
HUIHUI_RESTORE_MODEL = str(globals().get('HUIHUI_RESTORE_MODEL') or globals().get('model_path', ''))
HUIHUI_RESTORE_ALIAS = str(globals().get('MODEL_ALIAS', ''))
IMAGE_PORT = globals().get('IMAGE_PORT')
IMAGE_BASE_URL = globals().get('IMAGE_BASE_URL')
IMAGE_SERVER_PROCESS = globals().get('IMAGE_SERVER_PROCESS')
IMAGE_SERVER_LOG_HANDLE = globals().get('IMAGE_SERVER_LOG_HANDLE')

def _gpu_used_mib():
    try:
        out = subprocess.run(['nvidia-smi', '--query-gpu=memory.used', '--format=csv,noheader,nounits'], check=True, capture_output=True, text=True).stdout.strip().splitlines()[0]
        return int(float(out))
    except Exception as exc:
        raise RuntimeError('GPUメモリ使用量を確認できないため、画像モデルを起動しません。') from exc

def _match_huihui_argv(argv):
    if not argv:
        return False
    expected_bin = Path(str(globals().get('SERVER_BIN', ''))).resolve()
    try:
        actual_bin = Path(argv[0]).resolve()
    except Exception:
        return False
    if actual_bin != expected_bin:
        return False
    try:
        model_index = argv.index('-m')
        alias_index = argv.index('--alias')
    except ValueError:
        return False
    return (
        model_index + 1 < len(argv)
        and alias_index + 1 < len(argv)
        and Path(str(argv[model_index + 1])).resolve() == Path(HUIHUI_RESTORE_MODEL).resolve()
        and argv[alias_index + 1] == HUIHUI_RESTORE_ALIAS
    )

def _find_huihui_pid():
    try:
        import psutil
    except Exception as exc:
        raise RuntimeError('Huihuiの所有PIDを安全に確認するためpsutilが必要です。') from exc
    candidates = []
    for proc in psutil.process_iter(['pid', 'cmdline']):
        try:
            argv = proc.info.get('cmdline') or []
        except (psutil.NoSuchProcess, psutil.AccessDenied):
            continue
        if _match_huihui_argv(argv):
            candidates.append((proc.info['pid'], argv))
    if len(candidates) > 1:
        raise RuntimeError(f'Huihui所有候補PIDが複数あるため停止しません: {[x[0] for x in candidates]}')
    return candidates[0] if candidates else None

def stop_huihui_for_image():
    global server_process, server_log_handle
    found = _find_huihui_pid()
    if BASE_URL:
        health = requests.get(f'{BASE_URL}/health', timeout=10)
        health.raise_for_status()
        ids = [str(x.get('id')) for x in requests.get(f'{BASE_URL}/v1/models', timeout=10).json().get('data', [])]
        if MODEL_ALIAS not in ids:
            raise RuntimeError(f'Huihui alias確認失敗: expected={MODEL_ALIAS}, actual={ids}')
        slots = requests.get(f'{BASE_URL}/slots', timeout=10)
        if slots.ok:
            try:
                if any(bool(x.get('is_processing')) for x in slots.json()):
                    raise RuntimeError('Huihuiに処理中のリクエストがあるため、画像モデルを起動しません。')
            except ValueError:
                pass
    if server_process is not None and server_process.poll() is None:
        import psutil
        tracked_argv = psutil.Process(server_process.pid).cmdline()
        if not _match_huihui_argv(tracked_argv):
            raise RuntimeError('追跡中serverの実引数がHuihui model/aliasと一致しないため停止しません。')
        found = (server_process.pid, tracked_argv)
    if found is None:
        raise RuntimeError('Huihuiの自分のPID/model_path/aliasを確認できないため、GPUを解放しません。')
    if server_process is not None and server_process.poll() is None:
        server_process.terminate()
        try:
            server_process.wait(timeout=30)
        except subprocess.TimeoutExpired:
            server_process.kill(); server_process.wait(timeout=10)
    else:
        import psutil
        proc = psutil.Process(found[0]); proc.terminate(); proc.wait(timeout=30)
    server_process = None
    if server_log_handle is not None:
        server_log_handle.close(); server_log_handle = None
    time.sleep(3)
    used = _gpu_used_mib()
    if used > 2048:
        raise RuntimeError(f'Huihui停止後もGPU使用量が{used}MiBあるため、画像モデルを起動しません。')
    print(f'Huihuiの所有PID {found[0]} を停止し、GPU使用量 {used}MiB を確認しました。')

def _clear_image_process():
    global IMAGE_SERVER_PROCESS, IMAGE_SERVER_LOG_HANDLE, IMAGE_BASE_URL, IMAGE_PORT
    if IMAGE_SERVER_PROCESS is not None and IMAGE_SERVER_PROCESS.poll() is None:
        IMAGE_SERVER_PROCESS.terminate()
        try:
            IMAGE_SERVER_PROCESS.wait(timeout=30)
        except subprocess.TimeoutExpired:
            IMAGE_SERVER_PROCESS.kill(); IMAGE_SERVER_PROCESS.wait(timeout=10)
    IMAGE_SERVER_PROCESS = None
    IMAGE_BASE_URL = None
    IMAGE_PORT = None
    if IMAGE_SERVER_LOG_HANDLE is not None:
        IMAGE_SERVER_LOG_HANDLE.close()
    IMAGE_SERVER_LOG_HANDLE = None

def _image_runtime_preflight():
    global IMAGE_VENV, IMAGE_COMFY_DIR, IMAGE_HOST, PRIVATE_PUBLIC_KEY_PATH, venv_python
    image_venv = Path(str(globals().get('IMAGE_VENV') or '/content/qwen_image_work/venv_fixed'))
    comfy_dir = Path(str(globals().get('IMAGE_COMFY_DIR') or '/content/qwen_image_work/ComfyUI'))
    IMAGE_VENV = image_venv
    IMAGE_COMFY_DIR = comfy_dir
    IMAGE_HOST = str(globals().get('IMAGE_HOST') or '127.0.0.1')
    key_value = globals().get('PRIVATE_PUBLIC_KEY_PATH')
    if not key_value:
        raise RuntimeError('画像準備セルを先に実行してください。')
    PRIVATE_PUBLIC_KEY_PATH = Path(str(key_value))
    candidate = globals().get('venv_python')
    if not candidate or not Path(str(candidate)).is_file():
        candidate = image_venv / 'bin' / 'python'
        if not candidate.is_file():
            candidate = image_venv / 'Scripts' / 'python.exe'
    gguf_name = str(globals().get('IMAGE_GGUF') or 'qwen-image-2.1-UC-Q8_0.gguf')
    te_name = Path(str(globals().get('IMAGE_TE') or 'qwen3vl_8b_int8_convrot.safetensors')).name
    vae_name = Path(str(globals().get('IMAGE_VAE') or 'qwen_image_2.1_vae_bf16.safetensors')).name
    required = [
        Path(str(candidate)),
        comfy_dir / 'main.py',
        comfy_dir / 'custom_nodes' / 'ComfyUI-GGUF',
        comfy_dir / 'custom_nodes' / 'qwen_private_node' / '__init__.py',
        PRIVATE_PUBLIC_KEY_PATH,
        comfy_dir / 'models' / 'unet' / gguf_name,
        comfy_dir / 'models' / 'text_encoders' / te_name,
        comfy_dir / 'models' / 'vae' / vae_name,
        Path(str(globals().get('PRIVATE_RESULT_DIR') or '')),
    ]
    if not str(globals().get('PRIVATE_ROUTE_BASE_URL') or '').strip():
        raise RuntimeError('画像準備セルを先に実行してください。')
    if not str(globals().get('PRIVATE_ROUTE_TOKEN') or '').strip():
        raise RuntimeError('画像準備セルを先に実行してください。')
    if any(not path.exists() for path in required):
        raise RuntimeError('画像準備セルを先に実行してください。')
    venv_python = Path(str(candidate))
    return venv_python

def start_image_backend():
    global IMAGE_PORT, IMAGE_BASE_URL, IMAGE_SERVER_PROCESS, IMAGE_SERVER_LOG_HANDLE
    if IMAGE_SERVER_PROCESS is not None and IMAGE_SERVER_PROCESS.poll() is None and IMAGE_BASE_URL:
        try:
            if requests.get(f'{IMAGE_BASE_URL}/system_stats', timeout=3).status_code == 200:
                validate_image_object_info(IMAGE_BASE_URL)
                print('既存の検証済みComfyUIを再利用:', IMAGE_BASE_URL)
                return IMAGE_BASE_URL
        except Exception:
            _clear_image_process()
    _image_runtime_preflight()
    stop_huihui_for_image()
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as probe:
        probe.bind((IMAGE_HOST, 0)); IMAGE_PORT = probe.getsockname()[1]
    IMAGE_SERVER_LOG_HANDLE = None
    cmd = [str(venv_python), str(IMAGE_COMFY_DIR / 'main.py'), '--listen', IMAGE_HOST, '--port', str(IMAGE_PORT), '--disable-auto-launch', '--cache-none', '--preview-method', 'none', '--disable-metadata']
    IMAGE_SERVER_PROCESS = subprocess.Popen(cmd, cwd=IMAGE_COMFY_DIR, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, text=True)
    IMAGE_BASE_URL = f'http://{IMAGE_HOST}:{IMAGE_PORT}'
    deadline = time.monotonic() + 180
    while time.monotonic() < deadline:
        if IMAGE_SERVER_PROCESS.poll() is not None:
            _clear_image_process()
            raise RuntimeError('ComfyUIがヘルスチェック前に終了しました。')
        try:
            if requests.get(f'{IMAGE_BASE_URL}/system_stats', timeout=3).status_code == 200:
                validate_image_object_info(IMAGE_BASE_URL)
                print('ComfyUI起動OK:', IMAGE_BASE_URL)
                return IMAGE_BASE_URL
        except requests.RequestException:
            pass
        except Exception:
            _clear_image_process()
            raise
        time.sleep(2)
    _clear_image_process()
    raise TimeoutError('ComfyUIが180秒以内に起動しませんでした')

print('準備完了後に start_image_backend() を実行してください。実行するとHuihuiを停止し、画像GPUを占有します。')
```

```python
try:
    from google.colab import output as colab_output
    colab_output.enable_custom_widget_manager()
except ImportError:
    pass

import io, json, ipywidgets as widgets
from IPython.display import clear_output, display

def create_image_ui():
    prompt = widgets.Textarea(placeholder='画像生成または編集の指示', layout=widgets.Layout(width='100%', height='90px'))
    upload = widgets.FileUpload(accept='image/*', multiple=False)
    generate = widgets.Button(description='画像を生成', button_style='primary')
    edit = widgets.Button(description='画像を編集')
    download = widgets.Button(description='暗号化ファイルをPCへ保存')
    ack = widgets.Button(description='PCで復号・保存を確認して削除')
    ack_check = widgets.Checkbox(description='PCで復号画像を保存済み', value=False)
    back = widgets.Button(description='チャットに戻る')
    pending = widgets.Dropdown(options=(), description='暗号化結果')
    output = widgets.Output(layout={'border': '1px solid #ddd'})
    state = {'busy': False, 'last_result': None, 'last_ref': None}
    global IMAGE_LAST_OUTPUT
    IMAGE_LAST_OUTPUT = None

    def auth_headers():
        return {'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}

    def refresh_pending():
        try:
            entries = json.loads((PRIVATE_RESULT_DIR / 'manifest.json').read_text(encoding='utf-8'))
            pending.options = [(str(item['result_id']), str(item['result_id'])) for item in entries if isinstance(item, dict) and item.get('result_id')]
        except Exception:
            pending.options = ()

    def selected_metadata():
        result_id = pending.value or (state['last_result'] or {}).get('opaque_id')
        if not result_id:
            return None
        response = requests.get(f'{PRIVATE_ROUTE_BASE_URL}/private/result/{result_id}/metadata', headers=auth_headers(), timeout=30)
        response.raise_for_status()
        metadata = response.json()
        # Normalize route metadata to the UI/node schema.
        metadata['opaque_id'] = metadata.get('opaque_id', metadata.get('result_id'))
        metadata['blob_name'] = metadata.get('blob_name', metadata.get('blob'))
        return metadata

    pending.observe(lambda _: setattr(ack_check, 'value', False), names='value')

    def uploaded_ref():
        if not upload.value:
            return None
        item = next(iter(upload.value.values())) if isinstance(upload.value, dict) else upload.value[0]
        content = item.get('content', b'')
        if not str(item.get('type', '')).startswith('image/') or len(content) > 25 * 1024 * 1024:
            raise ValueError('25MiB以下の画像を選択してください。')
        try:
            from PIL import Image
            normalized = io.BytesIO()
            Image.open(io.BytesIO(content)).convert('RGB').save(normalized, format='PNG')
            content = normalized.getvalue()
        except Exception as exc:
            raise ValueError('画像をPNGへ変換できません。') from exc
        response = requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference', headers=auth_headers(), data=content, timeout=60)
        response.raise_for_status()
        ref_id = response.json()['ref_id']
        try: upload.value = ()
        except Exception: pass
        state['last_ref'] = ref_id
        return ref_id

    def show_result(metadata):
        global IMAGE_LAST_OUTPUT
        state['last_result'] = metadata
        IMAGE_LAST_OUTPUT = metadata
        ack_check.value = False
        refresh_pending(); pending.value = metadata.get('opaque_id')
        with output:
            clear_output(wait=True)
            print('暗号化結果を準備しました。opaque_id:', metadata.get('opaque_id'))
            print('PCへ暗号化ファイルを保存し、ローカルviewerで復号してください。')

    def run(graph):
        if state['busy']:
            return
        state['busy'] = True
        generate.disabled = True; edit.disabled = True; download.disabled = True; ack.disabled = True; back.disabled = True
        try:
            with output: clear_output(wait=True); print('生成中…')
            metadata, _ = wait_image_result(queue_image_graph(graph, base_url=IMAGE_BASE_URL), base_url=IMAGE_BASE_URL)
            show_result(metadata)
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass
            state['busy'] = False
            generate.disabled = False; edit.disabled = False; download.disabled = False; ack.disabled = False; back.disabled = False

    def on_generate(_):
        try:
            text = prompt.value.strip()
            if not text:
                with output: print('プロンプトを入力してください。')
                return
            run(build_t2i_graph(text))
        except Exception as exc:
            with output: print(f'画像生成エラー: {type(exc).__name__}')
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass

    def on_edit(_):
        try:
            text = prompt.value.strip()
            if not text:
                with output: print('編集指示を入力してください。')
                return
            ref_id = uploaded_ref() or state['last_ref']
            if not ref_id:
                raise ValueError('編集する画像をアップロードしてください。')
            run(build_i2i_graph(ref_id, text))
        except Exception as exc:
            with output: print(f'画像編集エラー: {type(exc).__name__}')
        finally:
            prompt.value = ''
            try: upload.value = ()
            except Exception: pass

    def on_download(_):
        try:
            metadata = selected_metadata()
            if not metadata: raise ValueError('保存する結果がありません。')
            from google.colab import files
            files.download(str(PRIVATE_RESULT_DIR / metadata['blob_name']))
            with output: print('暗号化ファイルの保存を開始しました。復号確認前に削除しません。')
        except Exception as exc:
            with output: print(f'暗号化ファイル保存エラー: {type(exc).__name__}')

    def on_ack(_):
        try:
            metadata = selected_metadata()
            if not metadata or not ack_check.value:
                with output: print('PCで復号画像を保存した後、確認欄を選択してください。')
                return
            response = requests.post(f"{PRIVATE_ROUTE_BASE_URL}/private/result/{metadata['opaque_id']}/ack", headers=auth_headers(), json={'recovered': True}, timeout=30)
            response.raise_for_status()
            state['last_result'] = None; IMAGE_LAST_OUTPUT = None; ack_check.value = False
            refresh_pending(); pending.value = None
            with output: clear_output(wait=True); print('回収確認済みの暗号化結果を削除しました。')
        except Exception as exc:
            with output: print(f'削除エラー: {type(exc).__name__}')

    def on_back(_):
        prompt.value = ''
        try: upload.value = ()
        except Exception: pass
        try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers=auth_headers(), json={}, timeout=10)
        except requests.RequestException: pass
        restore_huihui()

    generate.on_click(on_generate); edit.on_click(on_edit); download.on_click(on_download); ack.on_click(on_ack); back.on_click(on_back)
    refresh_pending()
    display(widgets.VBox([prompt, upload, pending, widgets.HBox([generate, edit, download, ack, back]), ack_check, output]))
    print('QWEN_PRIVATE_IMAGE_UI_READY', IMAGE_BASE_URL)

def stop_image_backend():
    _clear_image_process()
    try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers={'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}, json={}, timeout=10)
    except Exception: pass
    deadline = time.monotonic() + 60
    last_used = None
    while time.monotonic() < deadline:
        last_used = _gpu_used_mib()
        if last_used <= 2048: return
        time.sleep(2)
    raise RuntimeError('ComfyUI停止後もGPU使用量が残っています。')

def _huihui_health():
    base = str(globals().get('BASE_URL', '')).rstrip('/'); alias = str(globals().get('MODEL_ALIAS', ''))
    if not base or not alias: return False
    try:
        if requests.get(f'{base}/health', timeout=3).status_code != 200: return False
        ids = [str(x.get('id')) for x in requests.get(f'{base}/v1/models', timeout=3).json().get('data', [])]
        return alias in ids
    except requests.RequestException: return False

def _with_new_port(cmd):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as probe:
        probe.bind((SERVER_HOST, 0)); new_port = probe.getsockname()[1]
    out = list(cmd)
    if '--port' in out: out[out.index('--port') + 1] = str(new_port)
    else: out += ['--port', str(new_port)]
    return out, new_port

def restore_huihui():
    global server_cmd, server_process, server_log_handle, SERVER_PORT, BASE_URL
    if _huihui_health():
        try: requests.post(f'{PRIVATE_ROUTE_BASE_URL}/private/reference/clear', headers={'Authorization': f'Bearer {PRIVATE_ROUTE_TOKEN}'}, json={}, timeout=10)
        except Exception: pass
        chat_ui = create_chat_ui(BASE_URL, MODEL_ALIAS, initial_history=chat_history)
        display(widgets.VBox(list(chat_ui[:3]) + [chat_ui[3]]))
        print('HUIHUI_CHAT_REUSED', BASE_URL, MODEL_ALIAS)
        return
    stop_image_backend()
    if not HUIHUI_RESTORE_CMD or not HUIHUI_RESTORE_MODEL: raise RuntimeError('Huihuiの復帰情報がありません。')
    server_cmd, SERVER_PORT = _with_new_port(HUIHUI_RESTORE_CMD)
    if '--log-disable' not in server_cmd: server_cmd.append('--log-disable')
    BASE_URL = f'http://{SERVER_HOST}:{SERVER_PORT}'; server_log_handle = None
    server_process = subprocess.Popen(server_cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, text=True)
    deadline = time.monotonic() + 900
    while time.monotonic() < deadline:
        if server_process.poll() is not None: raise RuntimeError('Huihuiの復帰に失敗しました。')
        try:
            if requests.get(f'{BASE_URL}/health', timeout=3).status_code == 200: break
        except requests.RequestException: pass
        time.sleep(3)
    else: raise TimeoutError('Huihuiの復帰が制限時間内に完了しませんでした。')
    chat_ui = create_chat_ui(BASE_URL, MODEL_ALIAS, initial_history=chat_history)
    display(widgets.VBox(list(chat_ui[:3]) + [chat_ui[3]])); print('HUIHUI_CHAT_UI_RESTORED', BASE_URL, MODEL_ALIAS)

start_image_button = widgets.Button(description='画像モードに切替', button_style='warning')
restore_huihui_button = widgets.Button(description='チャットに戻る')
def _attempt_huihui_restore_after_image_failure():
    try:
        if not _huihui_health() and globals().get('HUIHUI_RESTORE_CMD') and globals().get('HUIHUI_RESTORE_MODEL'):
            restore_huihui()
    except Exception:
        pass

def _start_image(_):
    try:
        start_image_backend()
        create_image_ui()
    except Exception as exc:
        _attempt_huihui_restore_after_image_failure()
        print(f'画像backend開始エラー: {type(exc).__name__}')
def _restore_huihui(_):
    try: restore_huihui()
    except Exception as exc: print(f'Huihui復帰エラー: {type(exc).__name__}')
start_image_button.on_click(_start_image); restore_huihui_button.on_click(_restore_huihui)
display(widgets.HBox([start_image_button, restore_huihui_button]))
```
